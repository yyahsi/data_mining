# Recommendation Algorithms
The three recommendation models were evaluated using 5-fold cross-validation: Content-Based Filtering, Item-Based Collaborative Filtering, and Hybrid Recommendation.

Data Preparation

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Use the sampled review data
ratings_data = sampled_reviews.copy()

# Keep only the columns needed for recommendation
ratings_data = ratings_data[["user_id", "parent_asin", "rating"]]

# Remove missing values
ratings_data = ratings_data.dropna(subset=["user_id", "parent_asin", "rating"])

# Make sure rating is numeric
ratings_data["rating"] = pd.to_numeric(ratings_data["rating"], errors="coerce")
ratings_data = ratings_data.dropna(subset=["rating"])

print("Ratings data shape:", ratings_data.shape)
ratings_data.head()
```

## Baseline Model - Popularity Based  

```python
popular_items = (
    ratings_data.groupby("parent_asin")["rating"]
    .agg(["mean", "count"])
    .reset_index()
    .rename(columns={"mean": "average_rating", "count": "rating_count"})
)

# Keep only items with at least 5 ratings
popular_items = popular_items[popular_items["rating_count"] >= 5]

# Sort by average rating and number of ratings
popular_items = popular_items.sort_values(
    by=["average_rating", "rating_count"],
    ascending=False
)

popular_items.head(10)
```

```python
item_info = sampled_metadata.drop_duplicates("parent_asin").copy()

if "title" in item_info.columns:
    popular_items_with_titles = popular_items.merge(
        item_info[["parent_asin", "title"]],
        on="parent_asin",
        how="left"
    )
else:
    popular_items_with_titles = popular_items.copy()

popular_items_with_titles.head(10)
```

## Content Based Recommendation

```python
metadata_cb = sampled_metadata.copy()

text_columns = [
    "title",
    "features",
    "description",
    "categories",
    "details",
    "store"
]

metadata_cb["product_text"] = ""

for col in text_columns:
    if col in metadata_cb.columns:
        metadata_cb["product_text"] = (
            metadata_cb["product_text"] + " " + metadata_cb[col].fillna("").astype(str)
        )

metadata_cb = metadata_cb.drop_duplicates("parent_asin")
metadata_cb = metadata_cb[metadata_cb["parent_asin"].isin(ratings_data["parent_asin"].unique())]

print("Metadata for content-based model:", metadata_cb.shape)
metadata_cb[["parent_asin", "product_text"]].head()
```

### TF-IDF Conversion

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

tfidf_matrix = tfidf.fit_transform(metadata_cb["product_text"])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
item_indices = pd.Series(
    metadata_cb.index,
    index=metadata_cb["parent_asin"]
).drop_duplicates()
```

Recommending Similar Products

```python
example_item = metadata_cb["parent_asin"].iloc[0]
example_item
```

```python
item_position = metadata_cb.index.get_loc(
    metadata_cb[metadata_cb["parent_asin"] == example_item].index[0]
)

similarity_scores = cosine_similarity(
    tfidf_matrix[item_position],
    tfidf_matrix
).flatten()

similar_items = pd.DataFrame({
    "parent_asin": metadata_cb["parent_asin"].values,
    "similarity_score": similarity_scores
})

similar_items = similar_items[similar_items["parent_asin"] != example_item]
similar_items = similar_items.sort_values("similarity_score", ascending=False)

similar_items.head(10)
```

```python
if "title" in metadata_cb.columns:
    similar_items = similar_items.merge(
        metadata_cb[["parent_asin", "title"]],
        on="parent_asin",
        how="left"
    )

similar_items.head(10)
```

### Content Recommendation to a Target User

```python
target_user = ratings_data["user_id"].value_counts().index[0]
target_user
```

```python
user_liked_items = ratings_data[
    (ratings_data["user_id"] == target_user) &
    (ratings_data["rating"] >= 4)
]["parent_asin"].unique()

user_liked_items[:10]
```

```python
liked_item_positions = []

for item in user_liked_items:
    if item in metadata_cb["parent_asin"].values:
        position = metadata_cb.index.get_loc(
            metadata_cb[metadata_cb["parent_asin"] == item].index[0]
        )
        liked_item_positions.append(position)

liked_item_positions[:10]
```

```python
user_content_scores = cosine_similarity(
    tfidf_matrix[liked_item_positions],
    tfidf_matrix
).mean(axis=0)

content_recommendations = pd.DataFrame({
    "parent_asin": metadata_cb["parent_asin"].values,
    "content_score": user_content_scores
})

# Remove items the user already rated
already_rated = ratings_data[ratings_data["user_id"] == target_user]["parent_asin"].unique()

content_recommendations = content_recommendations[
    ~content_recommendations["parent_asin"].isin(already_rated)
]

content_recommendations = content_recommendations.sort_values(
    "content_score",
    ascending=False
)

if "title" in metadata_cb.columns:
    content_recommendations = content_recommendations.merge(
        metadata_cb[["parent_asin", "title"]],
        on="parent_asin",
        how="left"
    )

content_recommendations.head(10)
```

## Collobarative Filtering Recommender

```python
top_items = ratings_data["parent_asin"].value_counts().head(5000).index

cf_data = ratings_data[ratings_data["parent_asin"].isin(top_items)]

print("Collaborative filtering data shape:", cf_data.shape)
print("Unique users:", cf_data["user_id"].nunique())
print("Unique items:", cf_data["parent_asin"].nunique())
```

```python
user_item_matrix = cf_data.pivot_table(
    index="user_id",
    columns="parent_asin",
    values="rating"
).fillna(0)

print("User-item matrix shape:", user_item_matrix.shape)
user_item_matrix.head()
```

### Item Similarity Calculation

```python
item_similarity = cosine_similarity(user_item_matrix.T)

item_similarity_df = pd.DataFrame(
    item_similarity,
    index=user_item_matrix.columns,
    columns=user_item_matrix.columns
)

item_similarity_df.head()
```

```python
user_rating_summary = (
    cf_data.groupby("user_id")["rating"]
    .agg(["count", "nunique", "mean"])
    .reset_index()
)

candidate_users = user_rating_summary[
    (user_rating_summary["count"] >= 5) &
    (user_rating_summary["nunique"] >= 2)
]

candidate_users = candidate_users.sort_values("count", ascending=False)

candidate_users.head()
```

```python
# Select a target user
target_user = user_item_matrix.index[0]

print("Target user:", target_user)

user_ratings = user_item_matrix.loc[target_user]

# Items the user has already rated
rated_items = user_ratings[user_ratings > 0].index

print("User's ratings:")
print(user_ratings[rated_items].value_counts().sort_index())
```

```python
# Use items the user liked
liked_items = user_ratings[user_ratings >= 4].index

print("Number of liked items:", len(liked_items))
# Calculate recommendation score based on similarity to liked items
cf_scores = item_similarity_df[liked_items].sum(axis=1)

# Remove items the user already rated
cf_scores = cf_scores.drop(labels=rated_items, errors="ignore")

# Sort recommendations
cf_recommendations = cf_scores.sort_values(ascending=False).reset_index()

cf_recommendations.columns = ["parent_asin", "cf_score"]

cf_recommendations.head(10)
```

```python
if "title" in metadata_cb.columns:
    cf_recommendations = cf_recommendations.merge(
        metadata_cb[["parent_asin", "title"]],
        on="parent_asin",
        how="left"
    )

cf_recommendations.head(10)
```

## Hybrid Recommender

Score = α × ScoreCollobrativeFiltering + (1 − α) × ScoreContentBased

```python
alpha = 0.7
```

```python
hybrid_cf = cf_recommendations[["parent_asin", "cf_score"]].copy()

hybrid_cf.head()
```

```python
user_liked_items = ratings_data[
    (ratings_data["user_id"] == target_user) &
    (ratings_data["rating"] >= 4)
]["parent_asin"].unique()

liked_item_positions = []

for item in user_liked_items:
    if item in metadata_cb["parent_asin"].values:
        position = metadata_cb.index.get_loc(
            metadata_cb[metadata_cb["parent_asin"] == item].index[0]
        )
        liked_item_positions.append(position)

print("Liked items found in metadata:", len(liked_item_positions))
```

```python
user_content_scores = cosine_similarity(
    tfidf_matrix[liked_item_positions],
    tfidf_matrix
).mean(axis=0)

hybrid_cb = pd.DataFrame({
    "parent_asin": metadata_cb["parent_asin"].values,
    "content_score": user_content_scores
})

hybrid_cb.head()
```

### Combining Scores

```python
hybrid_scores = hybrid_cf.merge(
    hybrid_cb,
    on="parent_asin",
    how="inner"
)

hybrid_scores.head()
```

```python
hybrid_scores["cf_score_norm"] = (
    hybrid_scores["cf_score"] - hybrid_scores["cf_score"].min()
) / (
    hybrid_scores["cf_score"].max() - hybrid_scores["cf_score"].min()
)

hybrid_scores["content_score_norm"] = (
    hybrid_scores["content_score"] - hybrid_scores["content_score"].min()
) / (
    hybrid_scores["content_score"].max() - hybrid_scores["content_score"].min()
)
```

```python
alpha = 0.7

hybrid_scores["hybrid_score"] = (
    alpha * hybrid_scores["cf_score_norm"] +
    (1 - alpha) * hybrid_scores["content_score_norm"]
)

hybrid_scores = hybrid_scores.sort_values("hybrid_score", ascending=False)

hybrid_scores.head(10)
```

```python
if "title" in metadata_cb.columns:
    hybrid_scores = hybrid_scores.merge(
        metadata_cb[["parent_asin", "title"]],
        on="parent_asin",
        how="left"
    )

hybrid_scores.head(10)
```

Note: Collobarative Filtering as it is did not give satisfactory results since users tend to rate all games they liked with 5 and cf scores all become 5. Instead I used top-n recommendation, where items ranked according to their similarity to the products previously liked by the target users. I believe this still counts as a collobarative filtering. This way we have different cf scores and item recommendations become more meaningful.

## Feature Selection:

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_selection import SelectKBest, chi2, mutual_info_classif
```

```python
# Copy merged dataset
df_fs = df.copy()

# Create a binary target variable
# 1 = user liked the product, 0 = user did not strongly like it
df_fs["liked"] = (df_fs["rating"] >= 4).astype(int)

print(df_fs["liked"].value_counts())
```

```python
df_fs["product_text"] = ""

text_columns = [
    "title_meta",
    "title",
    "features",
    "description",
    "categories",
    "details",
    "store"
]

for col in text_columns:
    if col in df_fs.columns:
        df_fs["product_text"] = df_fs["product_text"] + " " + df_fs[col].fillna("").astype(str)

df_fs["product_text"].head()
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    min_df=2
)

X_text = tfidf.fit_transform(df_fs["product_text"])
y = df_fs["liked"]

feature_names = np.array(tfidf.get_feature_names_out())

print("Text feature matrix shape:", X_text.shape)
```

### Chi-Square Method

```python
chi2_selector = SelectKBest(score_func=chi2, k=50)

X_chi2 = chi2_selector.fit_transform(X_text, y)

chi2_scores = chi2_selector.scores_
chi2_selected_features = feature_names[chi2_selector.get_support()]

chi2_results = pd.DataFrame({
    "feature": chi2_selected_features,
    "chi2_score": chi2_scores[chi2_selector.get_support()]
}).sort_values("chi2_score", ascending=False)

chi2_results.head(20)
```

```python
chi2_results.to_csv("feature_selection_chi2.csv", index=False)
```

### Information Gain:

```python
mi_selector = SelectKBest(score_func=mutual_info_classif, k=50)

X_mi = mi_selector.fit_transform(X_text, y)

mi_scores = mi_selector.scores_
mi_selected_features = feature_names[mi_selector.get_support()]

mi_results = pd.DataFrame({
    "feature": mi_selected_features,
    "mutual_information_score": mi_scores[mi_selector.get_support()]
}).sort_values("mutual_information_score", ascending=False)

mi_results.head(20)
```

```python
mi_results.to_csv("feature_selection_mutual_information.csv", index=False)
```

## MAE and RMSE Evaluation

Here we compare the algorithms: Content-Based Filtering/
Item-Based Collaborative Filtering/
Hybrid Recommendation

### Recommender functions

Content Based:

```python
def predict_content_based(train_data, test_data, content_similarity_df):
    global_mean = train_data["rating"].mean()
    predictions = []

    for _, row in test_data.iterrows():
        user = row["user_id"]
        target_item = row["parent_asin"]

        user_history = train_data[train_data["user_id"] == user]

        if len(user_history) == 0:
            predictions.append(global_mean)
            continue

        if target_item not in content_similarity_df.index:
            predictions.append(global_mean)
            continue

        rated_items = user_history["parent_asin"].values
        ratings = user_history["rating"].values

        similarities = []

        for item in rated_items:
            if item in content_similarity_df.columns:
                similarities.append(content_similarity_df.loc[target_item, item])
            else:
                similarities.append(0)

        similarities = np.array(similarities)

        if similarities.sum() == 0:
            predictions.append(global_mean)
        else:
            predicted_rating = np.dot(similarities, ratings) / similarities.sum()
            predicted_rating = np.clip(predicted_rating, 1, 5)
            predictions.append(predicted_rating)

    return predictions
```

Collobarative Filtering:

```python
def predict_collaborative_filtering(train_data, test_data):
    global_mean = train_data["rating"].mean()

    user_item_matrix = train_data.pivot_table(
        index="user_id",
        columns="parent_asin",
        values="rating"
    ).fillna(0)

    item_similarity = cosine_similarity(user_item_matrix.T)

    item_similarity_df = pd.DataFrame(
        item_similarity,
        index=user_item_matrix.columns,
        columns=user_item_matrix.columns
    )

    predictions = []

    for _, row in test_data.iterrows():
        user = row["user_id"]
        target_item = row["parent_asin"]

        if user not in user_item_matrix.index:
            predictions.append(global_mean)
            continue

        if target_item not in item_similarity_df.index:
            predictions.append(global_mean)
            continue

        user_ratings = user_item_matrix.loc[user]
        rated_items = user_ratings[user_ratings > 0].index

        if len(rated_items) == 0:
            predictions.append(global_mean)
            continue

        similarities = item_similarity_df.loc[target_item, rated_items]
        ratings = user_ratings[rated_items]

        denominator = similarities.abs().sum()

        if denominator == 0:
            predictions.append(global_mean)
        else:
            predicted_rating = np.dot(similarities, ratings) / denominator
            predicted_rating = np.clip(predicted_rating, 1, 5)
            predictions.append(predicted_rating)

    return predictions
```

 Hybrid Filtering

```python
def predict_hybrid(cb_predictions, cf_predictions, alpha=0.7):
    hybrid_predictions = []

    for cb, cf in zip(cb_predictions, cf_predictions):
        hybrid_rating = alpha * cf + (1 - alpha) * cb
        hybrid_rating = np.clip(hybrid_rating, 1, 5)
        hybrid_predictions.append(hybrid_rating)

    return hybrid_predictions
```

### 5-Fold Validation 

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)

rating_results = []

fold_number = 1

for train_index, test_index in kf.split(eval_data):
    print("Running fold:", fold_number)

    train_data = eval_data.iloc[train_index].copy()
    test_data = eval_data.iloc[test_index].copy()

    # Keep evaluation manageable
    if len(test_data) > 3000:
        test_data = test_data.sample(3000, random_state=42)

    y_true = test_data["rating"].values

    # Content-Based
    cb_predictions = predict_content_based(
        train_data,
        test_data,
        content_similarity_df
    )

    cb_mae = mean_absolute_error(y_true, cb_predictions)
    cb_rmse = np.sqrt(mean_squared_error(y_true, cb_predictions))

    rating_results.append({
        "Fold": fold_number,
        "Model": "Content-Based Filtering",
        "MAE": cb_mae,
        "RMSE": cb_rmse
    })

    # Collaborative Filtering
    cf_predictions = predict_collaborative_filtering(
        train_data,
        test_data
    )

    cf_mae = mean_absolute_error(y_true, cf_predictions)
    cf_rmse = np.sqrt(mean_squared_error(y_true, cf_predictions))

    rating_results.append({
        "Fold": fold_number,
        "Model": "Item-Based Collaborative Filtering",
        "MAE": cf_mae,
        "RMSE": cf_rmse
    })

    # Hybrid
    hybrid_predictions = predict_hybrid(
        cb_predictions,
        cf_predictions,
        alpha=0.7
    )

    hybrid_mae = mean_absolute_error(y_true, hybrid_predictions)
    hybrid_rmse = np.sqrt(mean_squared_error(y_true, hybrid_predictions))

    rating_results.append({
        "Fold": fold_number,
        "Model": "Hybrid Recommendation",
        "MAE": hybrid_mae,
        "RMSE": hybrid_rmse
    })

    fold_number += 1
```

## Precision K - Recall K Evaluation

```python
K = 10
RELEVANCE_THRESHOLD = 4
def precision_at_k(recommended_items, relevant_items, k):
    recommended_k = recommended_items[:k]

    if len(recommended_k) == 0:
        return 0

    hits = len(set(recommended_k) & set(relevant_items))

    return hits / k
def recall_at_k(recommended_items, relevant_items, k):
    recommended_k = recommended_items[:k]

    if len(relevant_items) == 0:
        return 0

    hits = len(set(recommended_k) & set(relevant_items))

    return hits / len(relevant_items)
```

### Content-Based Top-K Recommender

```python
def recommend_content_based(train_data, user_id, content_similarity_df, k):
    user_history = train_data[train_data["user_id"] == user_id]

    if len(user_history) == 0:
        return []

    liked_items = user_history[user_history["rating"] >= 4]["parent_asin"].values
    seen_items = user_history["parent_asin"].values

    if len(liked_items) == 0:
        return []

    scores = pd.Series(0, index=content_similarity_df.index, dtype=float)

    for item in liked_items:
        if item in content_similarity_df.columns:
            scores = scores + content_similarity_df[item]

    scores = scores.drop(labels=seen_items, errors="ignore")

    recommended_items = scores.sort_values(ascending=False).head(k).index.tolist()

    return recommended_items
```

### Collobarative Top K

```python
def recommend_collaborative_filtering(train_data, user_id, k):
    user_item_matrix = train_data.pivot_table(
        index="user_id",
        columns="parent_asin",
        values="rating"
    ).fillna(0)

    if user_id not in user_item_matrix.index:
        return []

    item_similarity = cosine_similarity(user_item_matrix.T)

    item_similarity_df = pd.DataFrame(
        item_similarity,
        index=user_item_matrix.columns,
        columns=user_item_matrix.columns
    )

    user_ratings = user_item_matrix.loc[user_id]

    seen_items = user_ratings[user_ratings > 0].index
    liked_items = user_ratings[user_ratings >= 4].index

    if len(liked_items) == 0:
        return []

    cf_scores = item_similarity_df[liked_items].sum(axis=1)

    cf_scores = cf_scores.drop(labels=seen_items, errors="ignore")

    recommended_items = cf_scores.sort_values(ascending=False).head(k).index.tolist()

    return recommended_items
```

### Hybrid Top K

```python
def recommend_hybrid(train_data, user_id, content_similarity_df, k, alpha=0.7):
    user_item_matrix = train_data.pivot_table(
        index="user_id",
        columns="parent_asin",
        values="rating"
    ).fillna(0)

    if user_id not in user_item_matrix.index:
        return []

    user_ratings = user_item_matrix.loc[user_id]

    seen_items = user_ratings[user_ratings > 0].index
    liked_items = user_ratings[user_ratings >= 4].index

    if len(liked_items) == 0:
        return []

    # Collaborative filtering scores
    item_similarity = cosine_similarity(user_item_matrix.T)

    item_similarity_df = pd.DataFrame(
        item_similarity,
        index=user_item_matrix.columns,
        columns=user_item_matrix.columns
    )

    cf_scores = item_similarity_df[liked_items].sum(axis=1)
    cf_scores = cf_scores.drop(labels=seen_items, errors="ignore")

    # Content-based scores
    cb_scores = pd.Series(0, index=content_similarity_df.index, dtype=float)

    for item in liked_items:
        if item in content_similarity_df.columns:
            cb_scores = cb_scores + content_similarity_df[item]

    cb_scores = cb_scores.drop(labels=seen_items, errors="ignore")

    # Combine only common candidate items
    common_items = list(set(cf_scores.index) & set(cb_scores.index))

    if len(common_items) == 0:
        return []

    hybrid_scores = pd.DataFrame({
        "parent_asin": common_items,
        "cf_score": cf_scores.loc[common_items].values,
        "cb_score": cb_scores.loc[common_items].values
    })

    # Normalize scores
    if hybrid_scores["cf_score"].max() != hybrid_scores["cf_score"].min():
        hybrid_scores["cf_score_norm"] = (
            hybrid_scores["cf_score"] - hybrid_scores["cf_score"].min()
        ) / (
            hybrid_scores["cf_score"].max() - hybrid_scores["cf_score"].min()
        )
    else:
        hybrid_scores["cf_score_norm"] = 0

    if hybrid_scores["cb_score"].max() != hybrid_scores["cb_score"].min():
        hybrid_scores["cb_score_norm"] = (
            hybrid_scores["cb_score"] - hybrid_scores["cb_score"].min()
        ) / (
            hybrid_scores["cb_score"].max() - hybrid_scores["cb_score"].min()
        )
    else:
        hybrid_scores["cb_score_norm"] = 0

    hybrid_scores["hybrid_score"] = (
        alpha * hybrid_scores["cf_score_norm"] +
        (1 - alpha) * hybrid_scores["cb_score_norm"]
    )

    recommended_items = (
        hybrid_scores
        .sort_values("hybrid_score", ascending=False)
        .head(k)["parent_asin"]
        .tolist()
    )

    return recommended_items
```

### 5-Fold Cross Validation for Precision K and Recall K

```python
ranking_results = []

fold_number = 1

for train_index, test_index in kf.split(eval_data):
    print("Running ranking fold:", fold_number)

    train_data = eval_data.iloc[train_index].copy()
    test_data = eval_data.iloc[test_index].copy()

    train_users = set(train_data["user_id"].unique())
    test_users = set(test_data["user_id"].unique())

    common_users = list(train_users & test_users)

    # Limit users for runtime
    if len(common_users) > 200:
        common_users = common_users[:200]

    for user_id in common_users:
        relevant_items = test_data[
            (test_data["user_id"] == user_id) &
            (test_data["rating"] >= RELEVANCE_THRESHOLD)
        ]["parent_asin"].unique().tolist()

        if len(relevant_items) == 0:
            continue

        # Content-Based
        cb_recs = recommend_content_based(
            train_data,
            user_id,
            content_similarity_df,
            K
        )

        ranking_results.append({
            "Fold": fold_number,
            "Model": "Content-Based Filtering",
            "Precision@K": precision_at_k(cb_recs, relevant_items, K),
            "Recall@K": recall_at_k(cb_recs, relevant_items, K)
        })

        # Collaborative Filtering
        cf_recs = recommend_collaborative_filtering(
            train_data,
            user_id,
            K
        )

        ranking_results.append({
            "Fold": fold_number,
            "Model": "Item-Based Collaborative Filtering",
            "Precision@K": precision_at_k(cf_recs, relevant_items, K),
            "Recall@K": recall_at_k(cf_recs, relevant_items, K)
        })

        # Hybrid
        hybrid_recs = recommend_hybrid(
            train_data,
            user_id,
            content_similarity_df,
            K,
            alpha=0.7
        )

        ranking_results.append({
            "Fold": fold_number,
            "Model": "Hybrid Recommendation",
            "Precision@K": precision_at_k(hybrid_recs, relevant_items, K),
            "Recall@K": recall_at_k(hybrid_recs, relevant_items, K)
        })

    fold_number += 1
```

# Discussion

Observations:

- Content based has the best MAE, therefore predicting closest to ratings in absolute error. This means the item information like metadata and texts alone seem to create a good enough recommendation algorithm.


- Hybrid recommender has the lowest RMSE. Since RMSE penalizes errors more quadratically the hybrid model may be better at escaping very wrong prediction mistakes.


- Content based may miss products behaviorally related, two items maybe bought at the same time even though having completely different descriptions, whereas collobarative filtering can recognize such patterns. Howeber since the matrix is sparse, item- item similarity also becomes very weak. Therefore collobarative did not predict anything better than the other two.

- Hybrid Recommender has the highest  Precision K and Recall K. The Hybrid models improve the drawbacks stated above by using both signals therefore it has the valuable metrices like these the highest, as we observe %20 improvement.

Conclusion:

- Content based  may have the highest MAE but the other 3 metrices are the best at hybrid recommender algorithm. Thus we can safely claim Hybrid algorithm performs best in this scenario.
